```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow  # For displaying images in Colab

# Dictionary containing image paths and their corresponding titles
image_dict = {
    "/content/drive/MyDrive/soyabean/black/aug_img 10_0_2125.jpg": "Black Hilum",
    "/content/drive/MyDrive/soyabean/brown/aug_20250322_154226_0_1223.jpg": "Brown Hilum",
    "/content/drive/MyDrive/soyabean/brown/aug_20250322_154304_0_187.jpg": "Brown Hilum",
    "/content/drive/MyDrive/soyabean/unfit/aug_20250322_180601_0_2937.jpg": "Unfit Seed",
    "/content/drive/MyDrive/soyabean/unfit/aug_20250322_180701_0_6801.jpg": "Unfit Seed"
}

# Iterate through all images in the dictionary
for img_path, title in image_dict.items():
    # Load the image
    image = cv2.imread(img_path)

    if image is None:
        print(f"Error: Could not load {img_path}")
        continue  # Skip this image if it can't be loaded

    # Convert to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Apply Canny Edge Detection
    edges = cv2.Canny(gray, 100, 200)

    # Find contours
    contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    print(f"\nProcessing: {title} ({img_path})")

    for contour in contours:
        # Compute shape features
        area = cv2.contourArea(contour)
        perimeter = cv2.arcLength(contour, True)
        x, y, w, h = cv2.boundingRect(contour)
        aspect_ratio = float(w) / h if h != 0 else 0
        circularity = (4 * np.pi * area) / (perimeter ** 2) if perimeter != 0 else 0
        moments = cv2.moments(contour)
        hu_moments = cv2.HuMoments(moments).flatten()

        # Print extracted features
        print(f"  Area: {area}, Perimeter: {perimeter}, Aspect Ratio: {aspect_ratio}, Circularity: {circularity}")
        print(f"  Hu Moments: {hu_moments}")

        # Draw contours
        cv2.drawContours(image, [contour], -1, (0, 255, 0), 2)

    # Display results
    print(f"Displaying results for {title}...\n")
    print("=" * 50)

    cv2_imshow(image)  # Display the image with contours
    cv2_imshow(edges)  # Display the Canny edges

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow  # For Google Colab

# Load the image
image = cv2.imread("/content/drive/MyDrive/soyabean/black/aug_img 10_0_2125.jpg")

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)  # Show grayscale image

# Apply Canny Edge Detection
edges = cv2.Canny(gray, 100, 200)
cv2_imshow(edges)  # Show Canny edge-detected image

# Find contours
contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# Draw contours on a copy of the image
contour_img = image.copy()
cv2.drawContours(contour_img, contours, -1, (0, 255, 0), 2)
cv2_imshow(contour_img)  # Show image after finding boundaries

# Extract & print shape features
for contour in contours:
    area = cv2.contourArea(contour)
    perimeter = cv2.arcLength(contour, True)
    x, y, w, h = cv2.boundingRect(contour)
    aspect_ratio = float(w) / h if h != 0 else 0
    circularity = (4 * np.pi * area) / (perimeter ** 2) if perimeter != 0 else 0

    print(f"Area: {area}, Perimeter: {perimeter}, Aspect Ratio: {aspect_ratio}, Circularity: {circularity}")

# Show final image with extracted features
cv2_imshow(contour_img)
```

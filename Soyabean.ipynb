```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.feature import graycomatrix, graycoprops, local_binary_pattern
from scipy import ndimage

# Load the image in grayscale
image = cv2.imread("black1.jpg", cv2.IMREAD_GRAYSCALE)

# --- 1. Compute GLCM Features ---
glcm = graycomatrix(image, distances=[1], angles=[0], levels=256, symmetric=True, normed=True)
contrast = graycoprops(glcm, 'contrast')[0, 0]
energy = graycoprops(glcm, 'energy')[0, 0]
correlation = graycoprops(glcm, 'correlation')[0, 0]
homogeneity = graycoprops(glcm, 'homogeneity')[0, 0]

print(f"GLCM Features:\nContrast: {contrast:.4f}, Energy: {energy:.4f}, Correlation: {correlation:.4f}, Homogeneity: {homogeneity:.4f}")

# --- 2. Extract LBP Features ---
radius = 3
n_points = 8 * radius
lbp = local_binary_pattern(image, n_points, radius, method="uniform")

# --- 3. Apply Gabor Filters ---
def apply_gabor(image, theta):
    kernel = cv2.getGaborKernel((21, 21), 8.0, theta, 10.0, 0.5, 0, ktype=cv2.CV_32F)
    return cv2.filter2D(image, cv2.CV_8UC3, kernel)

theta_values = [0, np.pi/4, np.pi/2, 3*np.pi/4]  # Different orientations
gabor_images = [apply_gabor(image, theta) for theta in theta_values]

# --- Display Results ---
fig, axes = plt.subplots(2, 3, figsize=(12, 6))
axes[0, 0].imshow(image, cmap="gray")
axes[0, 0].set_title("Original Image")

axes[0, 1].imshow(lbp, cmap="gray")
axes[0, 1].set_title("LBP Features")

for i in range(4):
    ax = axes[1, i % 3]
    ax.imshow(gabor_images[i], cmap="gray")
    ax.set_title(f"Gabor (Theta={theta_values[i]:.2f})")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Step 1: Image Load Karo
image_path = "brown1.jpg"
img = cv2.imread(image_path)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)  # Grayscale convert

# Step 2: Thresholding (Hilum extract karne ke liye)
_, thresh = cv2.threshold(gray, 80, 255, cv2.THRESH_BINARY_INV)  # Adjust threshold if needed

# Step 3: Contours Find Karo (Hilum ka shape detect karne ke liye)
contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# Step 4: Contours Draw Karo
cv2.drawContours(img, contours, -1, (0,255,0), 2)  # Green Contours

# Step 5: Result Show Karo
cv2_imshow( img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
